### 📖 1. Dataset Overview

**MuMiN** (*Multilingual Multimodal Fact-Checked Misinformation*) is a large-scale, graph-based dataset designed for misinformation detection across social networks and news media.

Unlike standard tabular datasets, MuMiN is structured as a **heterogeneous graph**, connecting multiple distinct entity node types via relational edges:
* **Node Types**: `Claims`, `Tweets`, `Users`, `Articles`, `Replies`, and `Images`.
* **Edge Types**: `tweet_discusses_claim`, `article_discusses_claim`, `user_follows_user`, `user_retweeted_tweet`, `reply_quote_of_tweet`, etc.

This rich graph topology enables Graph Neural Network (GNN) architectures to model not only text and image contents but also user interaction networks, retweet cascades, and reply semantics.


### ⚠️ 2. Technical Constraints & Twitter/X API Bottlenecks

While MuMiN offers a unique multi-relational structure, utilizing it for end-to-end model training imposes major practical and technical constraints:

1. **Twitter/X API Bearer Token Requirement**:
   * For legal reasons and Terms of Service compliance, dataset authors only distribute tweet IDs and user IDs rather than raw text/media.
   * Compiling ("rehydrating") the dataset requires querying the Twitter/X API v2 with an active **Bearer Token**.
2. **API Access & Financial Cost**:
   * Following Twitter/X API v2 pricing changes, API rehydration requires enterprise/paid developer tiers ($500+/month).
   * Free/Basic tier rate limits make full dataset rehydration impractical (taking months or hitting strict quota limits).
3. **Link Rot & Media Unavailability**:
   * News article URLs and web-scraped image links associated with claims suffer from significant **link rot** (>60–90% dead links over time).
   * Fallback web archiving services (such as Wayback Machine) cannot retrieve many missing image assets.
4. **Class Imbalance**:
   * The dataset exhibits extreme class skew (~94.8% misinformation vs ~5.2% factual claims), requiring specialized sampling techniques for standard classification heads.

### 📌 4. Conclusion & Selection Decision

* **Theoretical Value**: MuMiN remains an important reference dataset for graph-based misinformation research.
